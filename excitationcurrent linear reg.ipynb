```python
from sklearn.preprocessing import StandardScaler
cols = ['Load_Current', 'Power_Factor', 'Pf_error']
scaler = StandardScaler()

df_cleaned[cols] = scaler.fit_transform(df_cleaned[cols])
```

```python
df_cleaned.head
```

```python
from scipy.stats import pearsonr
import pandas as pd

# List of features to check against target
selected_features = [
  'Load_Current', 'Power_Factor', 'Pf_error'
]

# Calculate Pearson correlation with target
results = []

for feature in selected_features:
    corr, p_value = pearsonr(df_cleaned[feature], df_cleaned[ 'Excitation_current'])
    results.append([feature, corr, p_value])

# Create DataFrame
corr_df = pd.DataFrame(
    results,
    columns=['Feature', 'Pearson Correlation', 'P-value']
)

# Sort by absolute correlation
corr_df = corr_df.sort_values(
    by='Pearson Correlation',
    key=lambda x: abs(x),
    ascending=False
)

print(corr_df)
```

```python
cat_features=[
    'Load_Current', 'Power_Factor', 'Pf_error'
]
    
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

df_cleaned['Excitation_current_bin'] = pd.qcut(df_cleaned['Excitation_current'], q=4, labels=False)
chi2_results = {}


for col in cat_features:
  contingency = pd.crosstab(df_cleaned[col], df_cleaned['Excitation_current_bin'])
  chi2_stat, p_val, _, _ = chi2_contingency(contingency)
  decision = (
      'Reject Null (Keep Feature)'
      if p_val < alpha
      else 'Accept Null (Drop Feature)'
  )

  chi2_results[col] = {
      'chi2_statistic': chi2_stat,
      'p_value': p_val, 
      'Decision': decision,
  }


chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')


chi2_df
```

```python
final_df=df_cleaned[['Load_Current', 'Power_Factor', 'Pf_error','Excitation_current']]
```

```python
final_df
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = final_df.drop('Excitation_current', axis=1)
y = final_df['Excitation_current']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
r2
n=x_test.shape[0]
p=x_test.shape[1]
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
adjusted_r2
```

```python
new_synchronous_dataset= pd.DataFrame([{
   	'Load_Current':6,
	'Power_Factor':0.99,
    'Pf_error':0.01
}])

new_synchronous_dataset[['Load_Current', 'Power_Factor', 'Pf_error']] = scaler.transform(new_synchronous_dataset[['Load_Current', 'Power_Factor', 'Pf_error']])

predicted_current_value = model.predict(new_synchronous_dataset)[0]
print('Predicted current value:', round(predicted_current_value, 2))
```
